```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression   
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.svm import SVR
import seaborn as sns
# Load dataset
task2_data = pd.read_csv('Task-Datasets/task2_svr_data.csv')
# Display first few rows of the dataset
display(task2_data.head())
```

```python
# Linear Regression (Without Scaling)
X = task2_data.iloc[:, 0:1].values
Y = task2_data.iloc[:, 1:2].values

lin_reg_NoScale = LinearRegression()
lin_reg_NoScale.fit(X, Y)

# Visualising Linear Regression (Without Scaling) results
plt.scatter(X,Y)
plt.plot(X, lin_reg_NoScale.predict(X), color='red')
plt.title('SVR Data (Linear Regression- Without Scaling)')
plt.xlabel('Temperature')
plt.ylabel('Ice Cream Sales')
plt.show()
```

```python
## SVR (Without Scaling)
# Fitting SVR to dataset
regressor = SVR(kernel='rbf')
regressor.fit(X, Y.ravel())

# Visualising SVR results
plt.scatter(X, Y)
plt.plot(X, regressor.predict(X), color='red')
plt.title('SVR Data (SVR model- Without Scaling')
plt.xlabel('Temperature')
plt.ylabel('Ice Cream Sales')
plt.show()
```

```python
#--- Predicting results---

#linear regression (Without scaling)
Y_pred_lin_NoScale = lin_reg_NoScale.predict([[20]])
```

```python
#svr 
Y_pred_svr_NoScale = regressor.predict(X=([[20]]))
```

```python
#printing results (Without Scaling)
print(f'Linear Regression Prediction (Without Scaling) for 20°C: {Y_pred_lin_NoScale}')
print(f'SVR Prediction (Without Scaling) for 20°C: {Y_pred_svr_NoScale}')
```

### OBSERVATION

The linear regression prediction at 20°C (281.25) which is very close to the observed value of (280 at 20°C) in the original dataset, indicating that the linear model provides a stable and reliable estimate in this region. 

In contrast, the large discrepancy observed in the Support Vector Regression (SVR) prediction arises from a model configuration issue rather than data behavior. SVR is highly sensitive to feature scaling, and when applied without scaling, the kernel-based distance calculations become distorted, leading to unreliable and inflated predictions. While linear regression is scale-invariant and therefore remains numerically stable, the unscaled SVR prediction of 423.06 is not meaningful and cannot be interpreted as a valid estimate.

```python
# SVR With Scaling
sc_X = StandardScaler()
sc_Y = StandardScaler()
X = sc_X.fit_transform(X)
Y = sc_Y.fit_transform(Y)
regressor = SVR(kernel='rbf')
regressor.fit(X, Y.ravel())
```

```python
# Visualising SVR (with Scaling) results
plt.scatter(X, Y)
plt.plot(X, regressor.predict(X), color='red')
plt.title('SVR Data (SVR model- With Scaling')
plt.xlabel('Temperature')
plt.ylabel('Ice Cream Sales')
```

```python
#--- Predicting results @ 27°C--- 

#linear regression (Without scaling)
Y_pred_lin_NoScale = lin_reg_NoScale.predict([[27]])
```

```python

# SVR (With Scaling)
Y_pred = regressor.predict(sc_X.transform([[27]]))
Y_pred_svr_Scaled = sc_Y.inverse_transform(Y_pred.reshape(-1, 1))
```

```python
#printing results SVR with and without scaling
print(f'SVR Prediction (Without Scaling) for 27°C: {Y_pred_svr_NoScale}')
print(f'SVR Prediction (With Scaling) for 27°C: {Y_pred_svr_Scaled}')
```

```python
#printing results LR (Without Scaling) and SVR (With Scaling)
print(f'Linear Regression Prediction (Without Scaling) for 27°C: {Y_pred_lin_NoScale}')
print(f'SVR Prediction (With Scaling) for 27°C: {Y_pred_svr_Scaled}')
```

### FINAL OBSERVATION 
At 27°C, the Linear Regression model (without scaling) predicts 512.12, which is above the actual dataset value of 480. This indicates that while the linear model captures the overall upward trend, it tends to overestimate at higher temperatures, reflecting the limitation of enforcing a single global linear relationship.

The Support Vector Regression (SVR) model without scaling predicts 423.06, which is significantly below the actual value. This large deviation highlights the impact of applying SVR without proper feature scaling, as the kernel distance calculations become distorted and produce unreliable results.

In contrast, the SVR model with scaling predicts 490.74, which is very close to the actual value of 480. This demonstrates that once scaling is applied, SVR is able to accurately capture the local non-linear structure of the data, providing a more reliable and interpretable estimate than both the unscaled SVR and the global linear regression.

### WHY FEATURE SCALING IS CRITICAL FOR SVR
Feature scaling is critical for SVR because the model relies on distance-based kernel functions; without scaling, these distance calculations become distorted, leading to unreliable predictions. With scaling applied, SVR is able to capture the non-linear pattern in the data and generalize effectively.”

```python
# Predicting results
#Linear Regression
Y_pred_lin = lin_reg.predict([[23]])
```

```python
# Decision Tree Regression
Y_pred_tree = regressor.predict([[23]])
```

```python
print(f'Linear Regression Prediction for 23 hours of Study: {Y_pred_lin}')
print(f'Decision Tree Regression Prediction for 23 hours of Study: {Y_pred_tree}')
```

### Why Decision tree is advantageous in this case 
Decision tree regression is advantageous for this dataset because it predicts based on local partitions rather than forcing a global linear trend. With study hours of 22 → 75 and 24 → 78, the tree assigns the score at 23 hours to the nearest observed value (75), reflecting the step‑like nature of exam scores. Linear regression, by contrast, smooths across all points and produces a fractional estimate (75.34), which does not fully capture the discrete jumps in the data. Thus, decision trees better respect local structure and discrete outcomes in small, non‑linear datasets.

## Model 3: Support Vector Regression
- Apply proper feature scaling (StandardScaler)

```python
# --- SUPPORT VECTOR REGRESSION---
# Load raw data
X = Assignment1_data[['Position_Level']].values
Y = Assignment1_data['Salary'].values

# Keep original data for inverse transformation later
X_original = X.copy()
Y_original = Y.copy()

# Feature Scaling
sc_X = StandardScaler()
sc_Y = StandardScaler()

X_scaled = sc_X.fit_transform(X_original)
Y_scaled = sc_Y.fit_transform(Y_original.reshape(-1, 1))

# Fitting SVR to dataset
svr_regressor = SVR(kernel='rbf')
svr_regressor.fit(X_scaled, Y_scaled.ravel())

# Visualising SVR results (Original Scale)
X_grid = np.linspace(X_original.min(), X_original.max(), 100).reshape(-1, 1)
Y_pred_scaled = svr_regressor.predict(sc_X.transform(X_grid))
Y_pred = sc_Y.inverse_transform(Y_pred_scaled.reshape(-1, 1))

plt.scatter(X_original, Y_original, color='blue')
plt.plot(X_grid, Y_pred, color='red')
plt.title('Position Level vs Salary (SVR Model)')
plt.xlabel('Position Level')
plt.ylabel('Salary')
plt.show()
```

```python
#--- Data Preparation
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression   
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
import seaborn as sns
#Loading the dataset
Assignment2_data = pd.read_csv('Assignment-Dataset/assignment2_energy_efficiency.csv')
Assignment2_data.head()
#check for null values
Assignment2_data.info()
#statistical summary
Assignment2_data.describe()

```

```python
X = Assignment2_data[['Temperature', 'Humidity', 'Wind_Speed', 'Solar_Radiation']].values
Y = Assignment2_data['Energy_Consumption'].values 

#splitting data into training and testing sets
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

2. Baseline: Multiple Linear Regression
- Build Multiple Linear Regression model
- Train on training set
- Make predictions on test set
- Calculate metrics:
  - R² score
  - Mean Absolute Error (MAE)
  - Root Mean Squared Error (RMSE)

```python
#---Model 1: Multiple Linear Regression---
lin_reg = LinearRegression()
lin_reg.fit(X_train, Y_train)
Y_pred_lin = lin_reg.predict(X_test)

# Calculate metrics
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
r2_lin = r2_score(Y_test, Y_pred_lin)
mae_lin = mean_absolute_error(Y_test, Y_pred_lin)
mse_lin = mean_squared_error(Y_test, Y_pred_lin)
rmse_lin = np.sqrt(mse_lin)
print(f'Multiple Linear Regression')
print(f'R²: {r2_lin}')
print(f'MAE: {mae_lin}')
print(f'MSE: {mse_lin}') # added it 
print(f'RMSE: {rmse_lin}')
```

3. Model 2: Support Vector Regression
- Apply StandardScaler to features
- Build SVR with RBF kernel
- Train and predict
- Calculate same metrics
- Compare with baseline

```python
# --- Model 2: Support Vector Regression ---

# 1. Feature Scaling (SVR is sensitive to feature scale)
sc_X = StandardScaler()
sc_Y = StandardScaler()

# Fit scaler on training data and transform
X_scaled_train = sc_X.fit_transform(X_train)
X_scaled_test = sc_X.transform(X_test)

# Scale target variable (SVR expects scaled targets for best performance)
Y_scaled_train = sc_Y.fit_transform(Y_train.reshape(-1, 1))

# 2. Build SVR model with RBF kernel
svr_reg = SVR(kernel='rbf')

# 3. Train the SVR model
svr_reg.fit(X_scaled_train, Y_scaled_train.ravel())

# 4. Make predictions on test data
Y_pred_scaled = svr_reg.predict(X_scaled_test)

# Inverse transform predictions to original scale
Y_pred_svr = sc_Y.inverse_transform(Y_pred_scaled.reshape(-1, 1))

# 5. Calculate evaluation metrics
r2_svr = r2_score(Y_test, Y_pred_svr)
mae_svr = mean_absolute_error(Y_test, Y_pred_svr)
mse_svr = mean_squared_error(Y_test, Y_pred_svr)
rmse_svr = np.sqrt(mse_svr)

# Print results for comparison with baseline
print("Model 2: Support Vector Regression")
print(f"R²: {r2_svr}")
print(f"MAE: {mae_svr}")
print(f"MSE: {mse_svr}")
print(f"RMSE: {rmse_svr}")
```

Model 3: Decision Tree Regression
- Build Decision Tree Regressor
- Try different max_depth values (3, 5, 10, None)
- For each depth:
  - Train and predict
  - Calculate metrics
- Identify best max_depth

```python
# --- Decision Tree Regression---
depths = [3, 5, 10, None]
results = {}
predictions = {}

for depth in depths:
    tree_reg = DecisionTreeRegressor(max_depth=depth, random_state=0)
    tree_reg.fit(X_train, Y_train)
    Y_pred_tree = tree_reg.predict(X_test)

    # Calculate metrics
    r2_tree = r2_score(Y_test, Y_pred_tree)
    mae_tree = mean_absolute_error(Y_test, Y_pred_tree)
    mse_tree = mean_squared_error(Y_test, Y_pred_tree)
    rmse_tree = np.sqrt(mse_tree)

    # Store results in dictionary 
    results[depth] = { 
    "R²": r2_tree, 
    "MAE": mae_tree, 
    "MSE": mse_tree, 
    "RMSE": rmse_tree 
    }
    predictions[depth] = Y_pred_tree # keep predictions for plotting

    # Print each depth’s metrics 
    print(f'Decision Tree Regression (max_depth={depth})') 
    print(f'R²: {r2_tree}') 
    print(f'MAE: {mae_tree}')
    print(f'MSE: {mse_tree}') 
    print(f'RMSE: {rmse_tree}\n')
```

Model Evaluation and Comparison
- Create comparison table with all metrics
- Visualize predictions vs actual values for each model:
  - Scatter plot (predicted vs actual)
  - Add diagonal line (perfect prediction)
- Create residual plots for each model

```python

# --- Find best depth (Highest R²) --- 
# `max(results, key=lambda d: results[d]["R²"])` loops through all depths stored in `results`, 
# # uses the lambda function to look up each depth’s R² value, 
# # and returns the depth with the maximum R² (best model performance).
best_depth = max(results, key=lambda d: results[d]["R²"]) 
best_metrics = results[best_depth] 
# retrieve predictions for best depth
Y_pred_tree_best = predictions[best_depth] 

print(f"Decision Tree Best depth is {best_depth} with R²={best_metrics['R²']:.3f}") 

# --- Build comparison table --- 
comparison_data = { "Model": ["Multiple Linear Regression", "Support Vector Regression", 
                              f"Decision Tree (depth={best_depth})"], 
                   "R²": [r2_lin, r2_svr, best_metrics["R²"]], 
                   "MAE": [mae_lin, mae_svr, best_metrics["MAE"]], 
                   "MSE": [mse_lin, mse_svr, best_metrics["MSE"]], 
                   "RMSE": [rmse_lin, rmse_svr, best_metrics["RMSE"]] 
                   } 
comparison_df = pd.DataFrame(comparison_data) 
comparison_df
```

#### 1.3 Data Preprocessing

**A. Handle Categorical Variables:**

  - Fuel_Type (OneHotEncoder)
  - Transmission (OneHotEncoder or LabelEncoder)
  - Accident_History (LabelEncoder: Yes=1, No=0)

**B. Train-Test Split:**

```python
# One-Hot Encoded features
OneHot_features = ['Brand', 'Fuel_Type', 'Transmission']

# Label Encoded features (as required)
LE_features = ['Accident_History', 'Service_Records']

# Numerical features
numerical_features = ['Year', 'Mileage', 'Engine_Size','Horsepower', 'Previous_Owners']

#LabelEncoding
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

for col in LE_features:
    assessment_data[col] = le.fit_transform(assessment_data[col])

# Column Transformer Building
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Define Column Transformer 
preprocessor = ColumnTransformer(
    transformers=[
        ('oneHot', OneHotEncoder(drop='first', handle_unknown='ignore'), OneHot_features),
        ('scale', StandardScaler(), numerical_features),
        ('le', 'passthrough', LE_features)
    ]
)

# Prepare feature matrix X
assessment_dataset = assessment_data[
    OneHot_features + numerical_features + LE_features
]
# Apply column transformer to feature matrix
assessment_transformed = preprocessor.fit_transform(assessment_dataset)

# Convert to Dataframe
feature_names = (
    preprocessor.named_transformers_['oneHot']
    .get_feature_names_out(OneHot_features)
)

all_feature_names = (
    list(feature_names)
    + numerical_features
    + LE_features
)

assessment_data_encoded = pd.DataFrame(assessment_transformed, columns=all_feature_names)
assessment_data_encoded.to_csv(
    'encoded_assessment_data.csv',
    index=False
)

display(assessment_data_encoded.head())        
assessment_data_encoded.info()
```

```python
from sklearn.linear_model import LinearRegression

model = Pipeline(steps=[
    ('preprocessing', preprocessor),
    ('regressor', LinearRegression())
])
#adding target variable to encoded data set
assessment_data_encoded["Price"] = assessment_data["Price"]

# Split data into features and target variable
X = assessment_data_encoded.drop('Price', axis=1)
Y = assessment_data_encoded['Price']

# Split data into training and testing sets
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=42)

#checks
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("Y_train shape:", Y_train.shape)
print("Y_test shape:", Y_test.shape)

```

### Phase 2: Model Development (Week 15 Skills)

#### 2.1 Baseline Model: Multiple Linear Regression
- Build Multiple Linear Regression model
- Train on training set
- Make predictions on both training and test sets
- Calculate evaluation metrics:
  - R² score (train and test)
  - Mean Absolute Error (MAE)
  - Mean Squared Error (MSE)
  - Root Mean Squared Error (RMSE)
- Store results for comparison

```python
# Baseline Model: Multiple Linear Regression
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
 
# Initialize and train modell
lin_reg = LinearRegression()
lin_reg.fit(X_train, Y_train)

# Predictions on training and test sets
Y_train_pred_lin = lin_reg.predict(X_train)
Y_test_pred_lin  = lin_reg.predict(X_test)


# Calculate evaluation metrics
# Training metrics 
r2_train = r2_score(Y_train, Y_train_pred) 
mae_train = mean_absolute_error(Y_train, Y_train_pred) 
mse_train = mean_squared_error(Y_train, Y_train_pred) 
rmse_train = np.sqrt(mse_train) 

# Test metrics 
r2_test = r2_score(Y_test, Y_test_pred)
mae_test = mean_absolute_error(Y_test, Y_test_pred)
mse_test = mean_squared_error(Y_test, Y_test_pred)
rmse_test = np.sqrt(mse_test)

print("Training Set:")
print("R²:", r2_train)
print("MAE:", mae_train)
print("MSE:", mse_train)
print("RMSE:", rmse_train)

print("\nTest Set:")
print("R²:", r2_test)
print("MAE:", mae_test)
print("MSE:", mse_test)
print("RMSE:", rmse_test)

# Store Results for Comparison 
MLR_results = {}
MLR_results["Multiple Linear Regression"] = {
    "Train R²": r2_train,
    "Test R²": r2_test,
    "Train MAE": mae_train,
    "Test MAE": mae_test,
    "Train MSE": mse_train,
    "Test MSE": mse_test,
    "Train RMSE": rmse_train,
    "Test RMSE": rmse_test
}
```

#### 2.2 Model 2: Polynomial Regression
- Create polynomial features (test degrees: 2, 3)
- For each degree:
  - Transform features
  - Train model
  - Calculate all metrics
  - Check for overfitting (compare train vs test R²)
- Select best polynomial degree
- **Note**: Be careful with high degrees - may cause overfitting

```python
# Model 2: Polynomial Regression

results_poly = {}

for degree in [2, 3]:
    # Create polynomial features
    poly = PolynomialFeatures(degree=degree)
    X_train_poly = poly.fit_transform(X_train)
    X_test_poly = poly.transform(X_test)
    
    # Train model
    poly_reg = LinearRegression()
    poly_reg.fit(X_train_poly, Y_train)
    
    # Predictions
    Y_train_pred = poly_reg.predict(X_train_poly)
    Y_test_pred = poly_reg.predict(X_test_poly)
    
    # Metrics
    r2_train = r2_score(Y_train, Y_train_pred)
    r2_test = r2_score(Y_test, Y_test_pred)
    mae_train = mean_absolute_error(Y_train, Y_train_pred)
    mae_test = mean_absolute_error(Y_test, Y_test_pred)
    mse_train = mean_squared_error(Y_train, Y_train_pred)
    mse_test = mean_squared_error(Y_test, Y_test_pred)
    rmse_train = np.sqrt(mse_train)
    rmse_test = np.sqrt(mse_test)
    
    # Store results
    results_poly[degree] = {
        "Train R²": r2_train,
        "Test R²": r2_test,
        "Train MAE": mae_train,
        "Test MAE": mae_test,
        "Train MSE": mse_train,
        "Test MSE": mse_test,
        "Train RMSE": rmse_train,
        "Test RMSE": rmse_test
    }
    
    # Print summary
    print(f"Polynomial Regression (degree={degree})")
    print("Train R²:", r2_train, "Test R²:", r2_test)
    print("Train MAE:", mae_train, "Test MAE:", mae_test)
    print("Train MSE:", mse_train, "Test MSE:", mse_test)
    print("Train RMSE:", rmse_train, "Test RMSE:", rmse_test)
    print("-"*50)

# Selecting best degree
best_degree = max(results_poly, key=lambda d: results_poly[d]["Test R²"])
print("Best polynomial degree based on metric result:", 'degree', best_degree)
```

#### 2.3 Model 3: Support Vector Regression
- Ensure features are properly scaled
- Build SVR with RBF kernel
- Try different parameters:
  - kernel='rbf', C=100, gamma='auto'
  - kernel='rbf', C=1000, gamma='scale'
- Train and evaluate each configuration
- Select best SVR model

```python
# Model 3: Support Vector Regression (SVR) with RBF Kernel
# I already infused a scaler in my pipeline
results_svr = {}

configs = [
    {"kernel": "rbf", "C": 100, "gamma": "auto"},
    {"kernel": "rbf", "C": 1000, "gamma": "scale"}
]

for cfg in configs:
    # Build model
    svr = SVR(kernel=cfg["kernel"], C=cfg["C"], gamma=cfg["gamma"])
    
    # Train
    svr.fit(X_train, Y_train)
    
    # Predictions
    Y_train_pred = svr.predict(X_train)
    Y_test_pred  = svr.predict(X_test)
    
    # Metrics
    r2_train = r2_score(Y_train, Y_train_pred)
    r2_test = r2_score(Y_test, Y_test_pred)
    mae_train = mean_absolute_error(Y_train, Y_train_pred)
    mae_test = mean_absolute_error(Y_test, Y_test_pred)
    mse_train = mean_squared_error(Y_train, Y_train_pred)
    mse_test = mean_squared_error(Y_test, Y_test_pred)
    rmse_train = np.sqrt(mse_train)
    rmse_test = np.sqrt(mse_test)
    
    # Store results
    key = f"SVR (C={cfg['C']}, gamma={cfg['gamma']})"
    results_svr[key] = {
        "Train R²": r2_train,
        "Test R²": r2_test,
        "Train MAE": mae_train,
        "Test MAE": mae_test,
        "Train MSE": mse_train,
        "Test MSE": mse_test,
        "Train RMSE": rmse_train,
        "Test RMSE": rmse_test
    }
    
    # Print summary
    print(key)
    print("Train R²:", r2_train, "Test R²:", r2_test)
    print("Train MAE:", mae_train, "Test MAE:", mae_test)
    print("Train MSE:", mse_train, "Test MSE:", mse_test)
    print("Train RMSE:", rmse_train, "Test RMSE:", rmse_test)
    print("-"*50)

# Select best SVR config
best_svr = max(results_svr, key=lambda k: results_svr[k]["Test R²"])
print("Best SVR configuration:", best_svr)
```

#### 2.4 Model 4: Decision Tree Regression
- Build Decision Tree Regressor
- Test different hyperparameters:
  - max_depth: 3, 5, 10, None
  - min_samples_split: 2, 5, 10
  - min_samples_leaf: 1, 2, 5
- For each configuration:
  - Train model
  - Calculate metrics
  - Check for overfitting
- Select best Decision Tree model
- Extract and visualize feature importances

```python
# Model 4: Decision Tree Regression with varying max depths and min samples splits/leafs
results_tree = {}
# Define Hyperparameter Grid
max_depths = [3, 5, 10, None]
min_samples_splits = [2, 5, 10]
min_samples_leafs = [1, 2, 5]

## Train and Evaluate Each Configuration
for depth in max_depths:
    for split in min_samples_splits: 
        for leaf in min_samples_leafs:
            tree_reg = DecisionTreeRegressor(max_depth=depth, min_samples_split=split, 
                                             min_samples_leaf=leaf, random_state=0)
            tree_reg.fit(X_train, Y_train)
            Y_train_pred = tree_reg.predict(X_train)
            Y_test_pred  = tree_reg.predict(X_test)
            
            r2_train = r2_score(Y_train, Y_train_pred)
            r2_test = r2_score(Y_test, Y_test_pred)
            mae_train = mean_absolute_error(Y_train, Y_train_pred)
            mae_test = mean_absolute_error(Y_test, Y_test_pred)
            mse_train = mean_squared_error(Y_train, Y_train_pred)
            mse_test = mean_squared_error(Y_test, Y_test_pred)
            rmse_train = np.sqrt(mse_train)
            rmse_test = np.sqrt(mse_test)
            
            #store results for comparison
            results_tree[(depth, split, leaf)] = {
                "Train R²": r2_train,
                "Test R²": r2_test,
                "Train MAE": mae_train,
                "Test MAE": mae_test,
                "Train MSE": mse_train,
                "Test MSE": mse_test,
                "Train RMSE": rmse_train,
                "Test RMSE": rmse_test,
            }
            
best_tree_config = max(results_tree, key=lambda d: results_tree[d]["Test R²"])
print("Best Decision Tree configuration based on Test R²:", best_tree_config)
print("Metrics:", results_tree[best_tree_config])

# Overfitting check
train_r2 = results_tree[best_tree_config]["Train R²"]
test_r2  = results_tree[best_tree_config]["Test R²"]
if train_r2 - test_r2 > 0.1:
    print("The best model may be overfitting!")
```
