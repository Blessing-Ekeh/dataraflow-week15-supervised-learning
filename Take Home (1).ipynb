```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression   
from sklearn.preprocessing import PolynomialFeatures
import seaborn as sns
# Load dataset
task1_data = pd.read_csv('Task-Datasets/task1_polynomial_data.csv')
# Display first few rows of the dataset
display(task1_data.head())
```

```python
X = task1_data.iloc[:, 0:1].values
Y = task1_data.iloc[:, 1].values
#visualize with scatter plots
plt.scatter (X,Y, color='red')
plt.title('Scatter plot of X vs Y', color='blue')
plt.xlabel('Experience Sales')
plt.ylabel('Salary')
```

```python
# Fitting Linear Regression to dataset
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
lin_reg.fit(X, Y)
```

```python
# Fitting Polynomial Regression deg = 2 to dataset
from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree=2) # Tranforming Tool
X_poly = poly_reg.fit_transform(X)
lin_reg2 = LinearRegression()
lin_reg2.fit(X_poly,Y)
```

```python
# Fitting Polynomial Regression deg = 3 to dataset
from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree=3) # Tranforming Tool
X_poly = poly_reg.fit_transform(X)
lin_reg3 = LinearRegression()
lin_reg3.fit(X_poly,Y)
```

```python
# Fitting Polynomial Regression deg = 4 to dataset
from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree=4) # Tranforming Tool
X_poly = poly_reg.fit_transform(X)
lin_reg4 = LinearRegression()
lin_reg4.fit(X_poly,Y)
```

```python
# Visualising Linear Regression results
plt.scatter(X,Y)
plt.plot(X, lin_reg.predict(X), color='red')
plt.title('Polynomial Data (Linear Regression)')
plt.xlabel('Experience Years')
plt.ylabel('Salary')
plt.show()
```

```python
# Visualising Polynomial Regression deg = 2 results
plt.scatter(X,Y) 
plt.plot(X, lin_reg2.predict(poly_reg.fit_transform(X)), color='red')
plt.title('Polynomial Data (Polynomial Regression Degree 2)')
plt.xlabel('Experience Years')
plt.ylabel('Salary')
plt.show()
```

```python
# Visualising Polynomial Regression deg = 3 results
plt.scatter(X,Y) 
plt.plot(X, lin_reg3.predict(poly_reg.fit_transform(X)), color='red')
plt.title('Polynomial Data (Polynomial Regression Degree 3)')
plt.xlabel('Experience Years')
plt.ylabel('Salary')
plt.show()
```

```python
# Visualising Polynomial Regression deg = 4 results
plt.scatter(X,Y) 
plt.plot(X, lin_reg4.predict(poly_reg.fit_transform(X)), color='red')
plt.title('Polynomial Data (Polynomial Regression Degree 4)')
plt.xlabel('Experience Years')
plt.ylabel('Salary')
plt.show()
```

```python
#--- COMPARISON OF PREDICTIONS ----#
# Predicting new result with Linear Regression
pred1 = lin_reg.predict([[8.5]])
```

```python
# Predicting new result with Polynomial Regression deg = 2
pred2 = lin_reg2.predict(poly_reg.fit_transform([[8.5]]))
```

```python
# Predicting new result with Polynomial Regression deg = 3
pred3 = lin_reg3.predict(poly_reg.fit_transform([[8.5]]))  
```

```python
# Predicting new result with Polynomial Regression deg = 4
pred4 = lin_reg4.predict(poly_reg.fit_transform([[8.5]]))
```

```python
print(f'Linear Regression Prediction for 8.5 years: {pred1}')
print(f'Polynomial Regression Degree 2 Prediction for 8.5 years: {pred2}')
print(f'Polynomial Regression Degree 3 Prediction for 8.5 years: {pred3}')
print(f'Polynomial Regression Degree 4 Prediction for 8.5 years: {pred4}')
```

From the above Result, Linear Regression is the Best model. 
### Explanation
 Although higher-degree polynomial models offer more flexibility, their predictions at 8.5 years are not significantly different from those of lower-degree polynomials or the linear regression model. While polynomial regression of degree 2 can capture non-linear relationships, the data and visualizations indicate that the relationship between experience and salary is largely linear. Given the increased risk of overfitting and higher variance with degree 2, 3, or 4 polynomials, linear regression provides the most reliable and generalizable performance, achieving an optimal balance between model complexity and predictive stability. Its simplicity, interpretability, and lower risk of overfitting make it the preferred choice.

### OBSERVATION

The linear regression prediction at 20°C (223.37) being far below the observed neighboring values (280 at 20 and 300 at 21) indicates that the model is underfitting the data. This suggests that the relationship between the variables is not globally linear, and that the linear model, which optimizes a single global trend, fails to capture the increased rate of change at higher input values. As a result, it does not respect the local structure of the data in this region.

In contrast, the large discrepancy observed in the Support Vector Regression (SVR) prediction arises from a model configuration issue rather than data behavior. SVR is highly sensitive to feature scaling, and when applied without scaling, the kernel-based distance calculations become distorted, leading to unreliable and inflated predictions. While linear regression is scale-invariant and therefore remains numerically stable, the unscaled SVR prediction of 423.06 is not meaningful and cannot be interpreted as a valid estimate.

```python
# SVR With Scaling
sc_X = StandardScaler()
sc_Y = StandardScaler()
X = sc_X.fit_transform(X)
Y = sc_Y.fit_transform(Y)
regressor = SVR(kernel='rbf')
regressor.fit(X, Y.ravel())
```

```python
# Visualising SVR (with Scaling) results
plt.scatter(X, Y)
plt.plot(X, regressor.predict(X), color='red')
plt.title('SVR Data (SVR model- With Scaling')
plt.xlabel('Temperature')
plt.ylabel('Ice Cream Sales')
```

```python
#--- Predicting results @ 27°C--- 

#linear regression (Without scaling)
Y_pred_lin_NoScale = lin_reg.predict([[27]])
```

```python

# SVR (With Scaling)
Y_pred = regressor.predict(sc_X.transform([[27]]))
Y_pred_svr_Scaled = sc_Y.inverse_transform(Y_pred.reshape(-1, 1))
```

```python
#printing results
print(f'Linear Regression Prediction (Without Scaling) for 27°C: {Y_pred_lin_NoScale}')
print(f'SVR Prediction (With Scaling) for 27°C: {Y_pred_svr_Scaled}')
```

### FINAL OBSERVATION 
Linear regression significantly underestimates ice-cream sales at higher temperatures, indicating that the relationship between temperature and sales is not globally linear. In contrast, Support Vector Regression, when properly scaled, produces a prediction very close to the observed value. 

### WHY FEATURE SCALING IS CRITICAL FOR SVR
Feature scaling is critical for SVR because the model relies on distance-based kernel functions; without scaling, these distance calculations become distorted, leading to unreliable predictions. With scaling applied, SVR is able to capture the non-linear pattern in the data and generalize effectively.”

### Task 3: Decision Tree Regression
**Objective**: Implement Decision Tree Regression and visualize decision boundaries

**Dataset**: `Task-Datasets/task3_decision_tree_data.csv`

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression   
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.svm import SVR
# Importing dataset
task3_data = pd.read_csv("Task-Datasets/task3_decision_tree_data.csv")

display(task3_data.head())
```

```python
X = task3_data.iloc[:, 0:1].values
Y = task3_data.iloc[:, 1:2].values
```

```python
# Fitting Decision Tree Regression to dataset
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor(random_state=0)
regressor.fit(X, Y)
```

```python
# Visualising Decision Tree Regression results (Standard Resolution)
plt.scatter(X,Y, color='red')
plt.plot(X, regressor.predict(X), color='blue')
plt.title('Decision Tree Regression Results (Standard Resolution)')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.show()

# Visualising Decision Tree Regression results (Higher resolution)
X_grid = np.arange(X.min(), X.max(), 0.01)
X_grid = X_grid.reshape((len(X_grid), 1))
plt.scatter(X, Y)
plt.plot(X_grid, regressor.predict(X_grid), color='red')
plt.title(' Decision Tree Regression Results (Higher Resolution)')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.show()
```

```python
#Linear Regression for comparison
lin_reg = LinearRegression()
lin_reg.fit(X, Y)
```

```python
# Visualizing both models for comparison
plt.scatter(X, Y, color='red')
plt.plot(X, lin_reg.predict(X), color='blue', label='Linear Regression')
plt.plot(X_grid, regressor.predict(X_grid), color='green', label='Decision Tree Regression')
plt.title('Comparison of Linear Regression and Decision Tree Regression')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.legend()
plt.show()
```

```python
# Predicting results
#Linear Regression
Y_pred_lin = lin_reg.predict([[23]])
```

```python
# Decision Tree Regression
Y_pred_tree = regressor.predict([[23]])
```

```python
print(f'Linear Regression Prediction for 23 hours of Study: {Y_pred_lin}')
print(f'Decision Tree Regression Prediction for 23 hours of Study: {Y_pred_tree}')
```

### Why Decision tree is advantageous in this case 
Decision tree regression is advantageous for this type of data because it makes predictions based on local partitions of the feature space rather than fitting a single global function. This allows the model to respect step-like changes and discrete outcomes in exam scores. In this case, the decision tree correctly predicts a score of 75 at 23 hours of study by aligning with nearby observed values, whereas linear regression produces a smoothed fractional value that does not fully reflect the local data structure.
